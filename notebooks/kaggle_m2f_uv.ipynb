```python
import os, sys, subprocess
from pathlib import Path

os.chdir("/kaggle/working")

try:
    from kaggle_secrets import UserSecretsClient
    sec = UserSecretsClient()
    for name in ("GIT_URL", "GITHUB_TOKEN"):
        try:
            os.environ[name] = sec.get_secret(name)
        except Exception:
            pass
except Exception:
    pass

REPO = Path("/kaggle/working/filament-segmentation-2026")
if REPO.exists():
    import shutil
    shutil.rmtree(REPO)

url = os.environ.get("GIT_URL", "")
token = os.environ.get("GITHUB_TOKEN", "")
if token and url.startswith("https://"):
    url = url.replace("https://", f"https://{token}@", 1)

subprocess.check_call(["git", "clone", "--depth", "1", "--branch", "refactor/logging", url, str(REPO)])

sys.path.insert(0, str(REPO))
os.chdir(REPO)
```
